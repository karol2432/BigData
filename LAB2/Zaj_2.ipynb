```
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import lit, when, col
```

**Zadanie 2**

```
display(dbutils.fs.ls("dbfs:/FileStore/tables/Files/"))
```

```
schema = StructType([
    StructField("id", StringType(), False),
    StructField("title", StringType(), False),
    StructField("original_title", StringType(), False),
    StructField("release_year", IntegerType(), False)
])

filePath = "dbfs:/FileStore/tables/Files/movies.csv"
df = spark.read.format("csv") \
            .option("header","true") \
            .schema(schema) \
            .load(filePath)

display(df)
```

**Zadanie 3**

```
df_filtered = df.filter(df.release_year.isNotNull())
```

```
incorrect_df = df.withColumn(
    "release_year", 
    when(col("release_year").cast("int") % 3 == 0, "string")  
    .otherwise(col("release_year")))
display(incorrect_df)
```

```
filePath = "dbfs:/FileStore/tables/Files/incorrect_mov.csv"
incorrect_df.write.format("csv").option("header", "true").mode("overwrite").save(filePath) 
```

```
permissive_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("mode", "PERMISSIVE") \
    .schema(schema) \
    .load(filePath)

display(permissive_df) 
```

```
failfast_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("mode", "FAILFAST") \
    .schema(schema) \
    .load(filePath)

display(failfast_df)
```

```
dropmalformed_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("mode", "DROPMALFORMED") \
    .schema(schema) \
    .load(filePath)

display(dropmalformed_df)
```

**Zadanie 4**

```
filePath = "dbfs:/FileStore/tables/Files/incorrect_mov.parquet"
incorrect_df.write.format("parquet").mode("overwrite").save(filePath)

new_df = spark.read.format("parquet").load(filePath)
display(new_df)
```

```
filePath = "dbfs:/FileStore/tables/Files/incorrect_mov.json"
incorrect_df.write.format("json").mode("overwrite").save(filePath)

new_df = spark.read.format("json").load(filePath)
display(new_df)
```
